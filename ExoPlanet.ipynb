```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
from astropy.timeseries import BoxLeastSquares
from scipy.signal import savgol_filter

# =========================
# 1. LOAD FITS DATA
# =========================

filename = "kplr010619192-2009166043257_llc.fits"

with fits.open(filename) as hdul:
    data = hdul[1].data
    time = np.array(data["TIME"], dtype=float)
    flux = np.array(data["PDCSAP_FLUX"], dtype=float)

# =========================
# 2. REMOVE INVALID VALUES
# =========================

mask = np.isfinite(time) & np.isfinite(flux)

time = time[mask]
flux = flux[mask]

# Remove zero/negative flux
mask = flux > 0

time = time[mask]
flux = flux[mask]

print("Number of data points:", len(time))

# =========================
# 3. NORMALIZE RAW FLUX
# =========================

flux = flux / np.nanmedian(flux)

# =========================
# 4. REMOVE LONG-TERM TREND
# =========================

# Sort by time
order = np.argsort(time)

time = time[order]
flux = flux[order]

# Smooth curve = stellar/instrumental long-term variation
window = 301

# Make window odd and smaller than data size
if window >= len(flux):
    window = len(flux) - 1

if window % 2 == 0:
    window -= 1

trend = savgol_filter(flux, window_length=window, polyorder=2)

# Detrended flux
detrended_flux = flux / trend

# Normalize again
detrended_flux = detrended_flux / np.median(detrended_flux)

# =========================
# 5. PLOT RAW vs DETRENDED
# =========================

plt.figure(figsize=(12,5))

plt.plot(time, flux, linewidth=0.5, label="Raw Flux")
plt.plot(time, trend, linewidth=1.5, label="Long-term Trend")

plt.xlabel("Time (days)")
plt.ylabel("Normalized Flux")
plt.title("Raw Kepler Light Curve + Trend")
plt.legend()
plt.show()


plt.figure(figsize=(12,5))

plt.plot(time, detrended_flux, ".", markersize=1)

plt.xlabel("Time (days)")
plt.ylabel("Detrended Flux")
plt.title("Detrended Kepler Light Curve")

plt.show()

# =========================
# 6. BLS SEARCH
# =========================

bls = BoxLeastSquares(time, detrended_flux)

# Search periods from 0.5 to 20 days
periods = np.linspace(0.5, 20, 10000)

# Transit durations
durations = np.array([0.02, 0.05, 0.10])

results = bls.power(
    periods,
    durations
)

# Best candidate
best_index = np.argmax(results.power)

best_period = results.period[best_index]
best_duration = results.duration[best_index]
best_power = results.power[best_index]

print("================================")
print("BLS TRANSIT SEARCH")
print("================================")

print(f"Best Period   : {best_period:.5f} days")
print(f"Transit Width : {best_duration:.5f} days")
print(f"BLS Power     : {best_power:.5f}")

# =========================
# 7. BLS PERIODOGRAM
# =========================

plt.figure(figsize=(12,5))

plt.plot(results.period, results.power)

plt.axvline(
    best_period,
    linestyle="--",
    label=f"Best Period = {best_period:.3f} days"
)

plt.xlabel("Orbital Period (days)")
plt.ylabel("BLS Power")
plt.title("BLS Transit Periodogram")
plt.legend()
plt.grid()

plt.show()

# =========================
# 8. PHASE FOLD
# =========================

phase = ((time - time[0]) / best_period) % 1

# Shift phase to -0.5 to +0.5
phase = phase - 0.5

# Sort phase
sort_phase = np.argsort(phase)

phase_sorted = phase[sort_phase]
flux_sorted = detrended_flux[sort_phase]

# =========================
# 9. PHASE-FOLDED LIGHT CURVE
# =========================

plt.figure(figsize=(12,5))

plt.plot(
    phase_sorted,
    flux_sorted,
    ".",
    markersize=2
)

plt.xlabel("Phase")
plt.ylabel("Normalized Flux")
plt.title("Phase-Folded Light Curve")

plt.ylim(
    np.percentile(flux_sorted, 1),
    np.percentile(flux_sorted, 99)
)

plt.grid()

plt.show()
```
